```python
# Import Libraries

import numpy as np
import networkx as nx
from scipy.spatial.distance import cosine
```

```python
# Create Vectors

np.random.seed(42)

vectors = np.random.rand(5, 4)

print(vectors)
```

```python
# Create HNSW Graph

graph = nx.Graph()

for i in range(len(vectors)):
    graph.add_node(i)

print(graph.nodes)
```

```python
# Add Connections

for i in range(len(vectors)):
    for j in range(i + 1, len(vectors)):

        # Use 1 - cosine distance to get cosine similarity
        similarity = 1 - cosine(
            vectors[i],
            vectors[j]
        )

        if similarity > 0.7:
            graph.add_edge(
                i,
                j,
                weight=similarity
            )
```

```python
# Query Vector

query = np.random.rand(4)

print("Query:", query)
```

```python
# Calculate Similarities

similarities = []

for i in range(len(vectors)):

    # Use 1 - cosine distance to get cosine similarity
    score = 1 - cosine(
        query,
        vectors[i]
    )

    similarities.append((i, score))
```

```python
# Rank Results

similarities.sort(
    key=lambda x: x[1],
    reverse=True
)

print(similarities)
```

```python
# Top Results

top_results = similarities[:3]

print("Top Results:")
print(top_results)
```

```python
# Results

print("Number of Vectors:", len(vectors))
print("Number of Connections:", graph.number_of_edges())
print("Top Similar Vectors:", top_results)
```
