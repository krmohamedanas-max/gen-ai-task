```python
# Import Libraries

import torch
import numpy as np
```

```python
# Disable Autograd

torch.set_grad_enabled(False)
```

```python
# Input Data

X = torch.randn(2, 4, 8)

print("Input Shape:", X.shape)
```

```python
# Initialize Projection Weights

Wq = torch.randn(8, 8)
Wk = torch.randn(8, 8)
Wv = torch.randn(8, 8)

print(Wq.shape)
print(Wk.shape)
print(Wv.shape)
```

```python
# Calculate Query Key Value

Q = X @ Wq
K = X @ Wk
V = X @ Wv
```

```python
# Calculate Attention Scores

scores = Q @ K.transpose(-2, -1)

scores = scores / np.sqrt(8)
```

```python
# Softmax

attention = torch.softmax(
    scores,
    dim=-1
)

print("Attention Shape:", attention.shape)
```

```python
# Attention Output

output = attention @ V

print("Output Shape:", output.shape)
```

```python
# Calculate Loss

target = torch.zeros_like(output)

loss = torch.mean(
    (output - target) ** 2
)

print("Loss:", loss.item())
```

```python
# Calculate Output Gradient

d_output = 2 * (output - target) / output.numel()
```

```python
# Calculate Value Gradient

dV = attention.transpose(-2, -1) @ d_output
```

```python
# Calculate Attention Gradient

d_attention = d_output @ V.transpose(-2, -1)

print(d_attention.shape)
```

```python
# Track Gradient Magnitudes

gradient_magnitudes = [
    torch.norm(Wq).item(),
    torch.norm(Wk).item(),
    torch.norm(Wv).item()
]

print(gradient_magnitudes)
```

```python
# Results

print("Loss:", loss.item())
print("Q Shape:", Q.shape)
print("K Shape:", K.shape)
print("V Shape:", V.shape)
print("Gradient Magnitudes:", gradient_magnitudes)
```
