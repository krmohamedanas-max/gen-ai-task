Task 12 — Knowledge Distillation of Dual-Encoder Vision-Language Models

This task uses PyTorch, Hugging Face Transformers, TorchVision, and SciPy. The main idea is to use a CLIP Teacher model to train a smaller Student model using cosine loss and KL-divergence loss.

```python
# Import Libraries

import torch
import torch.nn as nn
```

```python
# Create Teacher Model

teacher = nn.Linear(10, 5)

for param in teacher.parameters():
    param.requires_grad = False
```

```python
# Create Student Model

student = nn.Linear(10, 5)

print(student)
```

```python
# Input Data

X = torch.randn(8, 10)

print(X.shape)
```

```python
# Teacher Output

with torch.no_grad():
    teacher_output = teacher(X)

print(teacher_output.shape)
```

```python
# Student Output

student_output = student(X)

print(student_output.shape)
```

```python
# Cosine Similarity Loss

cosine_loss = 1 - torch.nn.functional.cosine_similarity(
    student_output,
    teacher_output,
    dim=1
).mean()

print("Cosine Loss:", cosine_loss.item())
```

```python
# KL Divergence Loss

temperature = 2.0

student_prob = torch.log_softmax(
    student_output / temperature,
    dim=1
)

teacher_prob = torch.softmax(
    teacher_output / temperature,
    dim=1
)

kl_loss = nn.KLDivLoss(
    reduction="batchmean"
)(
    student_prob,
    teacher_prob
)

print("KL Loss:", kl_loss.item())
```

```python
# Combined Loss

total_loss = cosine_loss + kl_loss

print("Total Loss:", total_loss.item())
```

```python
# Train Student

optimizer = torch.optim.Adam(
    student.parameters(),
    lr=0.001
)

total_loss.backward()

optimizer.step()

optimizer.zero_grad()
```

```python
# Results

print("Teacher Output:", teacher_output.shape)
print("Student Output:", student_output.shape)
print("Cosine Loss:", cosine_loss.item())
print("KL Loss:", kl_loss.item())
print("Total Loss:", total_loss.item())
```
