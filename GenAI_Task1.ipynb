```python
# Import NumPy

import numpy as np
```

```python
# Input Data

batch_size = 2
num_heads = 2
seq_len = 4
head_dim = 3

np.random.seed(42)

Q = np.random.randn(batch_size, num_heads, seq_len, head_dim)
K = np.random.randn(batch_size, num_heads, seq_len, head_dim)
V = np.random.randn(batch_size, num_heads, seq_len, head_dim)
```

```python
# Calculate Attention Scores

scores = np.matmul(Q, K.transpose(0, 1, 3, 2))
```

```python
# Scaling

scores = scores / np.sqrt(head_dim)
```

```python
# Causal Mask

mask = np.triu(np.ones((seq_len, seq_len)), k=1)

scores = np.where(mask == 1, -1e9, scores)
```

```python
# Softmax

def softmax(x):
    exp_x = np.exp(x - np.max(x, axis=-1, keepdims=True))
    return exp_x / np.sum(exp_x, axis=-1, keepdims=True)

attention_weights = softmax(scores)
```

```python
# Attention Output

attention_output = np.matmul(attention_weights, V)

print(attention_output)
```

```python
# Results

print("Q Shape:", Q.shape)
print("K Shape:", K.shape)
print("V Shape:", V.shape)
print("Output Shape:", attention_output.shape)
```
