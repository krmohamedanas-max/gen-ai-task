```python
# Import Libraries

import numpy as np
import torch
```

```python
# Training Text

text = "hello world hello machine learning"
```

```python
# Create Vocabulary

tokens = list(set(text.split()))

vocab = {token: i for i, token in enumerate(tokens)}

print(vocab)
```

```python
# Tokenize Text

token_ids = [vocab[token] for token in text.split()]

print(token_ids)
```

```python
# Create Token Pairs

pairs = []

for i in range(len(token_ids) - 1):
    pairs.append((token_ids[i], token_ids[i + 1]))

print(pairs)
```

```python
# Find Most Frequent Pair

from collections import Counter

pair_counts = Counter(pairs)

most_common_pair = pair_counts.most_common(1)[0]

print(most_common_pair)
```

```python
# Create Input and Target

X = torch.tensor(token_ids[:-1])
Y = torch.tensor(token_ids[1:])

print("Input:", X)
print("Target:", Y)
```

```python
# Causal Mask

seq_len = len(X)

mask = torch.triu(
    torch.ones(seq_len, seq_len),
    diagonal=1
)

mask = mask.masked_fill(mask == 1, float("-inf"))

print(mask)
```

```python
# Embedding Layer

embedding = torch.nn.Embedding(
    len(vocab),
    16
)

X_embed = embedding(X)

print(X_embed.shape)
```

```python
# Causal Language Model

model = torch.nn.Linear(
    16,
    len(vocab)
)

logits = model(X_embed)

print(logits.shape)
```

```python
# Loss Calculation

loss_function = torch.nn.CrossEntropyLoss()

loss = loss_function(
    logits,
    Y
)

print("Loss:", loss.item())
```

```python
# Results

print("Vocabulary:", vocab)
print("Token IDs:", token_ids)
print("Loss:", loss.item())
```
